```python
df = pd.read_csv("/home/milo/eternal_gallery/all.csv.gz", index_col=0)
```

```python
total_counts = (
   df.groupby([
            'disease_class',
            'sample'
        ])
        .count()
).iloc[:,[0]][
            df.groupby([
                    'disease_class',
                    'sample'
                ])
                .count()
                .iloc[:,0]
                .apply(lambda x: x != 0)
    ]

per_type_counts = (
    df.groupby([
            'level1_labels',
            'disease_class',
            'sample'
        ])
        .count()
        .iloc[:,[0]][
            df.groupby([
                    'level1_labels',
                    'disease_class', 
                    'sample'
                ])
                .count()
                .iloc[:,0]
                .apply(lambda x: x != 0)
    ]
)
per_type_counts.head()
```

```python
total_counts = pd.DataFrame(
   df.value_counts([
            'disease_class',
            'sample'
        ]))
per_type_counts = pd.DataFrame(
    df.value_counts([
            'level1_labels',
            'disease_class',
            'sample'
        ])
)
per_type_counts.head()
```

```python
merged_df = per_type_counts.join(total_counts, on=['disease_class', 'sample'], rsuffix="_total")
```

```python
merged_df['fraction'] = merged_df["count"]/merged_df["count_total"]
```
